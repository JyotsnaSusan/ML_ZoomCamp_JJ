```python
df_asia = df_car[df_car['origin'] == 'Asia']
x_df = df_asia[['vehicle_weight', 'model_year']]
```

```python
x_df=x_df.head(7)
```

```python
x_arr = x_df.to_numpy()
```

```python
XTX = x_arr.T @ x_arr
```

```python
XTX_Inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_Inv @ x_arr.T @ y
```
